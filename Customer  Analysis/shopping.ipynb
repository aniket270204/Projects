```python
pf["Review Rating"] = pf.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median())) 
```

```python
pf.isnull().sum()
```

```python
pf.columns = pf.columns.str.lower()
pf.columns = pf.columns.str.replace(' ','_')
pf = pf.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
pf.columns
```

```python
# create a colums age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
pf['age_group'] = pd.qcut(pf['age'], q=4, labels = labels)
```

```python
pf[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Quarterly': 30,
    "Bi-Weekly": 14,
    "Annually": 365,
    "Every 3 Months": 90
}

pf['purchase_frequency_days'] = pf["frequency_of_purchases"].map(frequency_mapping)
```

```python
pf[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
